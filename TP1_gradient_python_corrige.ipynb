## Partie 1 : fonction et gradient exact
On étudie $f(x, y) = x^2 + 3xy + y^2$ (déjà rencontrée en séance 2). Son gradient est $\nabla f(x, y) = (2x + 3y,\; 3x + 2y)^\top$.

**Exercice 1.1.** Écrire la fonction `f(x, y)` (elle doit marcher aussi bien avec des nombres qu'avec des tableaux `numpy`).

```python
def f(x, y):
    return x**2 + 3*x*y + y**2
```

**Exercice 1.2.** Écrire `grad_f(x, y)` qui renvoie le gradient sous la forme d'un tableau `numpy`.

```python
def grad_f(x, y):
    return np.array([2*x + 3*y, 3*x + 2*y])
```

```python
g = grad_f(1, 2)
print(g)                      # attendu : [8 7]
print(np.linalg.norm(g))      # attendu : sqrt(113), environ 10.63
```

## Partie 2 : gradient numérique (différences finies)
Quand on ne connaît pas la formule du gradient, on l'approche par des différences finies. Deux schémas :
- **avant** : $\dfrac{f(x+h, y) - f(x, y)}{h}$ ;
- **centrée** : $\dfrac{f(x+h, y) - f(x-h, y)}{2h}$ (et de même pour la variable $y$).

**Exercice 2.1.** Écrire `num_grad(func, x, y, h, schema)` qui renvoie le gradient approché ; `schema` vaut `"avant"` ou `"centree"`.

```python
def num_grad(func, x, y, h=1e-6, schema="centree"):
    if schema == "avant":
        dx = (func(x + h, y) - func(x, y)) / h
        dy = (func(x, y + h) - func(x, y)) / h
    else:
        dx = (func(x + h, y) - func(x - h, y)) / (2*h)
        dy = (func(x, y + h) - func(x, y - h)) / (2*h)
    return np.array([dx, dy])
```

```python
print("exact    :", grad_f(1, 2))
print("avant    :", num_grad(f, 1, 2, schema="avant"))
print("centrée  :", num_grad(f, 1, 2, schema="centree"))
```

**Exercice 2.2.** Étudier l'erreur en fonction du pas $h$ sur $g(x, y) = e^x \sin y$ au point $(1, 1)$ (gradient exact : $(e^x \sin y,\; e^x \cos y)$). Pour chaque $h$ de `hs`, calculer l'erreur (norme de la différence avec le gradient exact) pour les deux schémas, puis tracer les deux courbes en échelle logarithmique.

```python
def g(x, y): return np.exp(x) * np.sin(y)
def grad_g(x, y): return np.array([np.exp(x)*np.sin(y), np.exp(x)*np.cos(y)])

hs = np.logspace(-12, -1, 12)
exact = grad_g(1, 1)
err_avant = [np.linalg.norm(num_grad(g, 1, 1, h, "avant") - exact) for h in hs]
err_centree = [np.linalg.norm(num_grad(g, 1, 1, h, "centree") - exact) for h in hs]

plt.loglog(hs, err_avant, "o-", label="avant")
plt.loglog(hs, err_centree, "s-", label="centrée")
plt.xlabel("h"); plt.ylabel("erreur"); plt.legend(); plt.grid(True); plt.show()
```

**Question 2.3.** Pourquoi l'erreur ne diminue-t-elle pas indéfiniment quand $h \to 0$ ? Quel schéma est le plus précis et pour quelle valeur de $h$ ?

**Réponse :** Deux sources d'erreur : l'erreur de **troncature**, qui diminue avec $h$ (en $O(h)$ pour le schéma avant, en $O(h^2)$ pour le schéma centré), et l'erreur d'**arrondi**, qui augmente quand $h$ est très petit (on soustrait deux nombres presque égaux). La courbe a donc une forme en V. Le schéma centré est plus précis, avec un meilleur $h$ vers $10^{-6}$ à $10^{-5}$ (contre environ $10^{-8}$ pour le schéma avant).

## Partie 3 : visualiser une fonction de deux variables

**Exercice 3.4.** Au point $P = (1, 2)$, tracer sur les courbes de niveau le gradient (en rouge) et son opposé (en bleu), tous deux normalisés pour avoir une longueur 1.

## Partie 4 : variation selon différentes directions
Pour une direction unitaire $u = (\cos\theta, \sin\theta)$, la variation de $f$ au point $p$ dans la direction $u$ est la **dérivée directionnelle** $D_u f(p) = \nabla f(p) \cdot u$.

**Exercice 4.1.** Écrire `dir_deriv(p, theta)`, puis tracer $D_u f(p)$ pour $\theta \in [0, 2\pi]$ au point $p = (1,2)$. Retrouver numériquement la direction de plus forte augmentation et la comparer à celle du gradient (`np.arctan2`).

```python
def dir_deriv(p, theta):
    u = np.array([np.cos(theta), np.sin(theta)])
    return grad_f(*p) @ u

p = np.array([1.0, 2.0])
thetas = np.linspace(0, 2*np.pi, 361)
vals = np.array([dir_deriv(p, t) for t in thetas])
plt.plot(thetas, vals); plt.xlabel("θ (rad)"); plt.ylabel("D_u f(p)"); plt.grid(True); plt.show()

theta_max = thetas[np.argmax(vals)]
g = grad_f(*p)
print("θ du maximum    :", round(theta_max, 3))
print("θ du gradient   :", round(np.arctan2(g[1], g[0]), 3))
print("valeur maximale :", round(vals.max(), 3), " ; norme du gradient :", round(np.linalg.norm(g), 3))
```

**Question 4.2.** Quelle forme a la courbe ? Pour quelles directions la variation est-elle maximale, minimale, nulle ?

**Réponse :** La courbe est une sinusoïde $D_u f = \|\nabla f\| \cos(\theta - \theta_g)$, où $\theta_g$ est l'angle du gradient. Maximum $= \|\nabla f\|$ dans la direction du gradient ($\theta \approx 0{,}72$ rad), minimum $= -\|\nabla f\|$ dans la direction opposée, et variation nulle pour les deux directions perpendiculaires au gradient (tangentes à la courbe de niveau).

**Exercice 4.3.** Vérifier $D_u f(p)$ par différences finies : comparer $\dfrac{f(p + h u) - f(p)}{h}$ (avec $h = 10^{-6}$) et `dir_deriv(p, theta)` pour quelques angles.

```python
h = 1e-6
for theta in [0, np.pi/4, np.pi/2, np.pi, 3*np.pi/2]:
    u = np.array([np.cos(theta), np.sin(theta)])
    fd = (f(*(p + h*u)) - f(*p)) / h
    print(f"θ = {theta:5.2f} : exact = {dir_deriv(p, theta):8.4f} , différence finie = {fd:8.4f}")
```

**Exercice 4.4.** On se déplace d'une longueur $t$ à partir de $p$ dans trois directions unitaires : le gradient, l'opposé du gradient et la direction perpendiculaire $u_0 = (-g_2, g_1)/\|g\|$. Compléter le tableau des variations $f(p + t u) - f(p)$ pour $t = 10^{-1}, 10^{-2}, 10^{-3}$.

```python
g = grad_f(*p); ug = g / np.linalg.norm(g); u0 = np.array([-ug[1], ug[0]])
print("      t   |   +gradient   |  -gradient    |  perpendiculaire")
for t in [1e-1, 1e-2, 1e-3]:
    d = [f(*(p + t*u)) - f(*p) for u in (ug, -ug, u0)]
    print(f"  {t:7.0e} | {d[0]:13.6f} | {d[1]:13.6f} | {d[2]:13.6f}")
```

**Question 4.5.** Comment les trois colonnes évoluent-elles quand $t$ est divisé par 10 ? Qu'en déduisez-vous ?

**Réponse :** Pour $\pm$ gradient, la variation est divisée par environ 10 (elle est proportionnelle à $t$ : **premier ordre**, pente $\pm\|\nabla f\| \approx \pm 10{,}6$). Pour la direction perpendiculaire, elle est divisée par environ 100 (elle est en $t^2$ : **second ordre**). Le long de la courbe de niveau, $f$ ne varie quasiment pas au premier ordre. Dans la direction $-\nabla f$, $f$ diminue : c'est l'idée de la descente de gradient.

## Partie 5 : comparer plusieurs fonctions
On compare les champs de gradient de trois fonctions : $f_1 = x^2 + y^2$, $f_2 = x^2 - y^2$ et $f_3 = f$.

**Exercice 5.1.** Écrire les gradients de $f_1$ et $f_2$, puis tracer les trois champs de gradient côte à côte avec la fonction `trace_champ` fournie.

```python
def trace_champ(ax, func, grad, titre, lim=3, n=13):
    xs = np.linspace(-lim, lim, 200); X, Y = np.meshgrid(xs, xs)
    Xq, Yq = np.meshgrid(np.linspace(-lim, lim, n), np.linspace(-lim, lim, n))
    GX, GY = grad(Xq, Yq)
    ax.contour(X, Y, func(X, Y), levels=15, alpha=0.5)
    ax.quiver(Xq, Yq, GX, GY, color="red", angles="xy")
    ax.set_title(titre); ax.set_aspect("equal")

f1 = lambda x, y: x**2 + y**2
f2 = lambda x, y: x**2 - y**2
grad1 = lambda x, y: np.array([2*x, 2*y])
grad2 = lambda x, y: np.array([2*x, -2*y])

fig, axs = plt.subplots(1, 3, figsize=(15, 5))
trace_champ(axs[0], f1, grad1, "f1 = x² + y²")
trace_champ(axs[1], f2, grad2, "f2 = x² − y²")
trace_champ(axs[2], f, grad_f, "f3 = x² + 3xy + y²")
plt.show()
```

**Question 5.2.** Les trois fonctions ont un gradient nul en $(0,0)$. D'après les courbes de niveau et les flèches, ce point est-il un minimum dans chaque cas ? Comment le voit-on ? (La classification rigoureuse sera faite avec la hessienne à la séance suivante.)

**Réponse :** Pour $f_1$, les flèches pointent toutes vers l'extérieur et les courbes de niveau sont des cercles autour de $(0,0)$ : c'est un **minimum**. Pour $f_2$ et $f_3$, les courbes de niveau sont des hyperboles : les flèches pointent vers l'extérieur dans une direction et vers l'intérieur dans l'autre, et $f$ diminue dans certaines directions : c'est un **point selle**, pas un minimum. Un gradient nul ne suffit donc pas à conclure.
